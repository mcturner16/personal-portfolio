```python
# Import seaborn
import seaborn as sns

# Apply the default theme
sns.set_theme(context="paper", font_scale=1.5, style="whitegrid")

# Load an example dataset
tips = sns.load_dataset("tips")

# Create a visualization
g = sns.relplot(
    data=tips,
    x="total_bill", y="tip",
    hue="time", style="smoker", size="size",
    aspect=1.5,
    alpha=0.75,
    sizes=(20,120)
)

g.set_axis_labels(
    "Total Bill (USD)",
    "Tip (USD)"
)
# Some ideas to try (you don't need all of them):
#
# Changes to set_theme...
#sns.set_theme(context=...)    #-> preset text and line sizes: "paper", "notebook", "talk", "poster"
#sns.set_theme(font_scale=...) #-> scales all text up or down together
#sns.set_theme(style=...)      #-> background and grid: "white", "whitegrid", "ticks", "dark", ...
#
# Changes within sns.replot...
# relplot(height=...)           -> height of the figure in inches
# relplot(aspect=...)           -> width as a multiple of the height
# relplot(sizes=(min, max))     -> smallest and largest marker size for the "size" variable
# relplot(alpha=...)            -> marker transparency, helps where points overlap
# relplot(palette=...)          -> colors used for the "hue" variable
# relplot(legend=...)           -> "auto", "brief", "full", or False to hide the legend
#
# Saved figure options...
# g = sns.relplot(...)          -> save the result so you can adjust it afterward:
# g.set_axis_labels("x label", "y label")    -> rename the axes
# sns.move_legend(g, loc=..., bbox_to_anchor=...)   -> move the legend
# g.tight_layout()              -> removes extra white space around the plot
#
# g.savefig("tiny.png", dpi=...)  -> save an image; dpi sets how many pixels per inch
```
