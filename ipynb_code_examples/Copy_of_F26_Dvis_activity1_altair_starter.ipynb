```python
import altair as alt
from vega_datasets import data

source = data.cars()      # Please, do not use this exact dataset for your first portfolio assignment
source.head()
```

## 2. Build one chart
Some ways to make it interactive:
- `tooltip=[...]` &rarr; hover tooltips
- `.interactive()` &rarr; pan / zoom


```python
chart = (
    alt.Chart(source)
    .mark_circle(size=60, opacity=0.7)
    .encode(
        x=alt.X("Horsepower:Q"), #Q for quantitative
        y=alt.Y("Miles_per_Gallon:Q", title="Miles per gallon"),
        color=alt.Color("Origin:N"), #N for nominal (categorical)
        tooltip=["Name:N", "Horsepower:Q", "Miles_per_Gallon:Q"],
    )
    .properties(width=600, height=360, title="Fuel efficiency vs. horsepower")
    .interactive()
)
chart
```

## 3. Get the spec &rarr; paste into a webpage
Copy **everything** this prints and paste it in `post.html` as `const spec1 = <paste>;`. The data is included in the spec for this example, so nothing else needs to be copied over.

```python
print(chart.to_json())
```
