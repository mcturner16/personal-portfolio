# Activity 6: Maps
Load map shapes, join data to them, and compare color choices.

Run the cells in order. This notebook needs an internet connection to load the map shapes.

````python
import altair as alt

GEO   = 'https://cdn.jsdelivr.net/npm/vega-datasets@v1.29.0/data/us-10m.json'
UNEMP = 'https://cdn.jsdelivr.net/npm/vega-datasets@v1.29.0/data/unemployment.tsv'

us_states   = alt.topo_feature(GEO, 'states')
us_counties = alt.topo_feature(GEO, 'counties')
````

## 1. Draw the shapes
`mark_geoshape` draws the geometry and `.project()` sets the projection. No data yet, just outlines.

````python
alt.Chart(us_states).mark_geoshape(fill='#e8e8e8', stroke='white').project(
    'albersUsa').properties(width=520, height=320, title='US states')
````

## 2. Join data to the shapes
`transform_lookup` matches each county's `id` to the unemployment table, then `color` encodes the rate.

````python
alt.Chart(us_counties).mark_geoshape().encode(
    color=alt.Color('rate:Q', scale=alt.Scale(scheme='blues'), title='Unemployment')
).transform_lookup(
    lookup='id', from_=alt.LookupData(UNEMP, 'id', ['rate'])
).project('albersUsa').properties(width=520, height=320, title='County unemployment')
````

## 3. Change the binning
Copy the map above and add `type='quantile'` to the scale:
`scale=alt.Scale(scheme='blues', type='quantile')`

Compare the two maps. Which areas look different?

````python
# your quantile map

alt.Chart(us_counties).mark_geoshape().encode(
    color=alt.Color('rate:Q', scale=alt.Scale(scheme='blues', type='quantile'), title='Unemployment')
).transform_lookup(
    lookup='id', from_=alt.LookupData(UNEMP, 'id', ['rate'])
).project('albersUsa').properties(width=520, height=320, title='County unemployment')
````

## 4. Try a diverging scale
Rates are stored as fractions (0.08 means 8%). Change the scale to a diverging scheme with a midpoint:
`scale=alt.Scale(scheme='redblue', reverse=True, domainMid=0.06)`

Then try `domainMid=.10`. Compare the two.

````python
# your diverging map

alt.Chart(us_counties).mark_geoshape().encode(
    color=alt.Color('rate:Q', scale=alt.Scale(scheme='redblue', reverse=True, domainMid=0.10), title='Unemployment')
).transform_lookup(
    lookup='id', from_=alt.LookupData(UNEMP, 'id', ['rate'])
).project('albersUsa').properties(width=520, height=320, title='County unemployment')
````

## 5. Zoom to Michigan
Filter to Michigan counties and switch the projection by adding:
```python
.transform_filter('datum.id >= 26000 && datum.id < 27000').project('mercator')
```

````python
# your Michigan map

alt.Chart(us_counties).mark_geoshape().encode(
    color=alt.Color('rate:Q', scale=alt.Scale(scheme='redblue', reverse=True, domainMid=0.10), title='Unemployment')
).transform_lookup(
    lookup='id', from_=alt.LookupData(UNEMP, 'id', ['rate'])
).project('albersUsa').properties(width=520, height=320, title='County unemployment'
).transform_filter('datum.id >= 26000 && datum.id < 27000').project('mercator')
````
