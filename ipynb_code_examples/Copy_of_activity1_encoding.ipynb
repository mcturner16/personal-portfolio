### 1. Start simple
One bar per city, but this collapses all 12 months into one number. What do we lose?

```python
alt.Chart(df).mark_bar().encode(x='city:N', y='mean(avg_high_f):Q')
```

### 2. Group two ways
Below is the **seasonal curve** (month on x, one line per city).

Then **# YOUR TURN:** make the *other* grouping — cities side by side within each month (a bar chart with month on the x-axis, color based on city, and you may want to include: `xOffset='city'`). Which one answers *“how do the cities compare in July?”* and which answers *“what’s the shape of the year?”*

```python
alt.Chart(df).mark_line(point=True).encode(
    x=alt.X('month:N', sort=months), y='avg_high_f:Q', color='city:N')
```

```python
# YOUR TURN: grouped bars (cities within each month)
alt.Chart(df).mark_bar().encode(
    x=alt.X('month:N', sort=months), y='avg_high_f:Q', color='city:N', xOffset='city:N')
```

### 3. The whole matrix at once

```python
# YOUR TURN: heatmap
alt.Chart(df).mark_rect().encode(x=alt.X('month:N', sort=months), y='city:N', color=alt.Color('avg_high_f:Q', scale=alt.Scale(scheme='blueorange')))


```
